```python
import spacy
```

```python
nlp =spacy.load("en_core_web_sm")
text = "West Chesterfieldville was referenced in Mr. Deeds."
```

```python
doc = nlp(text)
```

Entity Ruler

```python
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
ruler = nlp.add_pipe("entity_ruler")
```

```python
nlp.analyze_pipes()
```

```python
patterns = [
    {"label": "GPE", "pattern": "West Chesterfieldville"}
]
```

```python
ruler.add_patterns(patterns)
```

```python
doc2 = nlp(text)
for ent in doc2.ents:
    print(ent.text, ent.label_)
```

```python
nlp2 = spacy.load("en_core_web_sm")
ruler = nlp2.add_pipe("entity_ruler", before="ner")
ruler.add_patterns(patterns)
doc = nlp2(text)
```

```python
for ent in doc.ents:
    print(ent.text,ent.label_)
```

```python
nlp3 = spacy.load("en_core_web_sm")
```

```python
ruler = nlp3.add_pipe("entity_ruler",before="ner")

patterns = [
    {"label": "GPE", "pattern": "West Chesterfieldville"},
    {"label": "FILM", "pattern": "Mr.Deeds"}
]
```

```python
ruler.add_patterns(patterns)
doc = nlp3(text)

for ent in doc.ents:
    print(ent.text, ent.label_)
```

Mr. Deeds could be a person or a film. Which is called a Toponym

Toponym resolution is when resolving this issue
