```python
import spacy
import numpy as np
```

```python
nlp = spacy.load("en_core_web_md")
with open ("data/wiki_us.txt", "r") as f:
    text = f.read()
doc = nlp(text)
sentence_1 = list(doc.sents)[0]
```

```python
print(sentence_1)
```

```python
your_word = "country"

ms = nlp.vocab.vectors.most_similar(
    np.asarray([nlp.vocab.vectors[nlp.vocab.strings[your_word]]]), n=10)
words = [nlp.vocab.strings[w] for w in ms[0][0]]
distances = ms[2]
print(words)
```

```python
doc_1 = nlp("I like salty fries and hamburgers.")
doc_2 = nlp("Fast food tastes very good.")
print(doc_1, "<->",doc_2,doc_1.similarity(doc_2))
```
