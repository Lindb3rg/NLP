```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
with open ("data/wiki_us.txt", "r") as f:
    text = f.read()
```

```python
print(text)
```

```python
doc = nlp(text)
```

```python
print(doc)
```

```python
print(len(text))
print(len(doc))
```

```python
for token in doc[0:10]:
    print(token)
```

```python
for sent in doc.sents:
    print(sent)
```
